```python
# Merge Issues and Billing data
risk_df = issues.merge(billing, on='customer_id')
```

```python
# Function to calculate customer risk score
def calculate_risk(row):
    score = 0

    # High resolution time
    if row['resolution_time_hrs'] > 24:
        score += 40

    # Delayed payment
    if row['payment_status'] == 'Delayed':
        score += 30

    # Unresolved issue
    if row['resolved'] == 'No':
        score += 30

    return score

# Apply function
risk_df['risk_score'] = risk_df.apply(calculate_risk, axis=1)
```

```python
# Create Risk Category
risk_df['risk_category'] = risk_df['risk_score'].apply(
    lambda x: 'High Risk' if x >= 60
    else ('Medium Risk' if x >= 30 else 'Low Risk')
)
```

```python
# View results
risk_df[['customer_id', 'risk_score', 'risk_category']].head()
```

```python
# Count customers by risk category
risk_df['risk_category'].value_counts()
```

###Which cities require urgent network improvement?

```python
# Complaints by city
city_issues = df.groupby('city')['issue_id'].count().sort_values(ascending=False)

print(city_issues)
```

Chennai recorded the highest number of network complaints (71,113), followed by Delhi (67,441) and Bangalore (63,089). These cities should be prioritized for network infrastructure improvements and proactive monitoring. High complaint volumes indicate potential network quality issues that may negatively impact customer satisfaction and increase churn risk.

Does poor network quality affect payments and churn?

```python
# Average resolution time by payment status
df.groupby('payment_status')['resolution_time_hrs'].mean()
```

Customers with delayed payments experienced longer issue resolution times compared to customers who paid on time. This suggests that poor network quality and slow complaint resolution may reduce customer satisfaction, leading to payment delays and increasing the risk of customer churn.

Are 5G users experiencing fewer issues?

```python
# Issues count by network type
df.groupby('network_type')['issue_id'].count()
```

4G users reported significantly more network issues (252,674) compared to 5G users (112,968). This indicates that 5G networks are providing a more reliable and stable user experience. The lower complaint volume among 5G users suggests that newer network infrastructure is helping reduce connectivity and service-related problems.

Which customer type generates the highest data usage?

```python
usage_customer = usage.merge(
    customers[['customer_id','customer_type']],
    on='customer_id'
)

usage_customer.groupby('customer_type')['data_used_gb'].mean().sort_values(ascending=False)
```

Postpaid customers consume slightly more data on average (2.482 GB) than prepaid customers (2.474 GB). Although the difference is small, postpaid users appear to be marginally more engaged with telecom services and may contribute higher long-term revenue.
